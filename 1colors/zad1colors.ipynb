```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

N = 1000
points = np.random.rand(N, 3)
colors = points.copy()

plt.figure(figsize=(8, 6))
plt.scatter(points[:, 0], points[:, 1], c=colors, s=15)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Projection XY (Matplotlib)")
plt.grid(True)
plt.savefig("plot_2d.png")
plt.show()

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(points[:, 0], points[:, 1], points[:, 2], c=colors, s=20)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title("3D Projection (Matplotlib)")
plt.savefig("plot_3d_static.png")
plt.show()

plotly_colors = ['rgb({},{},{})'.format(int(r*255), int(g*255), int(b*255)) for r, g, b in colors]
fig_plotly = go.Figure(data=[go.Scatter3d(
    x=points[:, 0], y=points[:, 1], z=points[:, 2],
    mode='markers',
    marker=dict(size=3, color=plotly_colors)
)])
fig_plotly.update_layout(title='3D Visualization (Plotly)')
fig_plotly.show()


center = np.array([0.5, 0.5, 0.5])
dist = np.linalg.norm(points - center, axis=1)
norm_dist = (dist - dist.min()) / (dist.max() - dist.min())
colors_alt = plt.cm.viridis(norm_dist)[:, :3]

fig_alt = plt.figure(figsize=(8, 8))
ax_alt = fig_alt.add_subplot(111, projection='3d')
ax_alt.scatter(points[:, 0], points[:, 1], points[:, 2], c=colors_alt, s=20)
ax_alt.set_title("Alternative Coloring (Distance to Center)")
plt.savefig("plot_3d_alt.png")
plt.show()
```
